### Exercise 1

Find the alphabet of a sequence, i.e. the set of distinct symbols it is made of, and compute the relative frequency (percentage) of each symbol.
Input: S = ACGTTGCATTACGTTGCAGT

```python
def alphabet(seq):
    return sorted(set(seq))


def relative_frequencies(seq):
    return {letter: seq.count(letter) / len(seq) for letter in alphabet(seq)}


S = "ACGTTGCATTACGTTGCAGT"
print(f"Sequence: {S}")
print(f"Length = {len(S)}")
print(f"Alphabet: {', '.join(alphabet(S))}")
print()
print("Absolute / relative frequencies:")
for letter, freq in relative_frequencies(S).items():
    print(f"{letter}: {S.count(letter):2d}  ->  {freq:.2f} ({freq * 100:.0f}%)")
```
