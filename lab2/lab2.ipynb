The melting temperature(Tm) is the temp at which one half of a particular DNA duplex will dissociate and become a simple strand of DNA. 
Primer length and sequence are of critical importance i designing the parameters of a successful amplific. 
The Tm of a nucleic asset duplex increases both with its length and with an increased C.G. content. 
A simple formula for calculating. Of Tm is Tm=4(G+C)+2(A+T). 
Make a simple application in any environment or program. 
Language which is able to calculate the Tm based on an input sequence S=ATCGCGTA

```python
def melting_temperature(seq):
    seq = seq.upper()
    invalid = set(seq) - set("ACGT")
    if invalid:
        raise ValueError(f"Invalid nucleotides: {invalid}")
    gc = seq.count("G") + seq.count("C")
    at = seq.count("A") + seq.count("T")
    return 4 * gc + 2 * at


S = "ATCGCGTA"
print(f"Sequence: {S}")
print(f"G+C = {S.count('G') + S.count('C')}, A+T = {S.count('A') + S.count('T')}")
print(f"Tm = {melting_temperature(S)} °C")
```
