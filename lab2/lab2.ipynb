### Exercise 1

The melting temperature (Tm) is the temperature at which one half of a particular DNA duplex will dissociate and become a single strand of DNA.
Primer length and sequence are of critical importance in designing the parameters of a successful amplification.
The Tm of a nucleic acid duplex increases both with its length and with an increased G+C content.
A simple formula for calculating Tm is: Tm = 4(G+C) + 2(A+T).

Make a simple application, in any environment or programming language, which is able to calculate the Tm based on an input sequence: S = ATCGCGTA

```python
def melting_temperature(seq):
    seq = seq.upper()
    invalid = set(seq) - set("ACGT")
    if invalid:
        raise ValueError(f"Invalid nucleotides: {invalid}")
    gc = seq.count("G") + seq.count("C")
    at = seq.count("A") + seq.count("T")
    return 4 * gc + 2 * at


S = "ATCGCGTA"
print(f"Sequence: {S}")
print(f"G+C = {S.count('G') + S.count('C')}, A+T = {S.count('A') + S.count('T')}")
print(f"Tm = {melting_temperature(S)} °C")
```

### Exercise 2

The actual Tm is influenced by the concentration of Mg, K and co-solvents. An alternative formula is:
Tm = 81.5 + 16.6 log10([Na+]) + 0.41(%GC) - 600/length

Implement an app that calculates the Tm of a DNA sequence using one of these formulas or both.
Input: S = ACGCGTGCCA

```python
import math


def gc_percent(seq):
    seq = seq.upper()
    return 100 * (seq.count("G") + seq.count("C")) / len(seq)


def melting_temperature_salt(seq, na=0.05):
    """Salt-adjusted Tm; na = [Na+] in mol/L."""
    seq = seq.upper()
    invalid = set(seq) - set("ACGT")
    if invalid:
        raise ValueError(f"Invalid nucleotides: {invalid}")
    return 81.5 + 16.6 * math.log10(na) + 0.41 * gc_percent(seq) - 600 / len(seq)


S = "ACGCGTGCCA"
NA = 0.05  # 50 mM

print(f"Sequence: {S}")
print(f"Length = {len(S)}, %GC = {gc_percent(S):.1f}%, [Na+] = {NA} M")
print()
print(f"Formula 1 (basic):         Tm = {melting_temperature(S)} °C")
print(f"Formula 2 (salt-adjusted): Tm = {melting_temperature_salt(S, NA):.2f} °C")
```
