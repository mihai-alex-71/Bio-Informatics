# Lab 2 — DNA Melting Temperature (Tm)

The **melting temperature (Tm)** is the temperature at which **half** of a DNA duplex
dissociates into single strands.

- Primer **length** and **sequence** are critical when designing a successful PCR amplification.
- Tm **increases** with the length of the duplex and with its **G+C content**
  (G–C pairs have 3 hydrogen bonds, A–T pairs only 2).

## Exercise 1 — Basic formula (Wallace rule)

$$T_m = 4\,(G + C) + 2\,(A + T)$$

| Symbol | Meaning |
|---|---|
| $G + C$ | number of G and C nucleotides |
| $A + T$ | number of A and T nucleotides |

**Input:** `S = ATCGCGTA`

**Task:** write a simple application that calculates Tm for the input sequence.

```python
def melting_temperature(seq):
    seq = seq.upper()
    invalid = set(seq) - set("ACGT")
    if invalid:
        raise ValueError(f"Invalid nucleotides: {invalid}")
    gc = seq.count("G") + seq.count("C")
    at = seq.count("A") + seq.count("T")
    return 4 * gc + 2 * at


S = "ATCGCGTA"
print(f"Sequence: {S}")
print(f"G+C = {S.count('G') + S.count('C')}, A+T = {S.count('A') + S.count('T')}")
print(f"Tm = {melting_temperature(S)} °C")
```

## Exercise 2 — Salt-adjusted formula

The real Tm is also influenced by the concentration of **Mg²⁺**, **K⁺** and **co-solvents**.
An alternative formula that takes the salt concentration into account is:

$$T_m = 81.5 + 16.6\,\log_{10}\!\left([\mathrm{Na^+}]\right) + 0.41\,(\%GC) - \frac{600}{\text{length}}$$

| Symbol | Meaning |
|---|---|
| $[\mathrm{Na^+}]$ | sodium ion concentration, in **mol/L** (here: 50 mM = 0.05 M, a typical PCR value) |
| $\%GC$ | percentage of G and C nucleotides in the sequence (0–100) |
| length | number of nucleotides in the sequence |

**Input:** `S = ACGCGTGCCA`

**Task:** write an application that calculates Tm for the input sequence using one or both formulas.

```python
import math


def gc_percent(seq):
    seq = seq.upper()
    return 100 * (seq.count("G") + seq.count("C")) / len(seq)


def melting_temperature_salt(seq, na=0.05):
    """Salt-adjusted Tm; na = [Na+] in mol/L."""
    seq = seq.upper()
    invalid = set(seq) - set("ACGT")
    if invalid:
        raise ValueError(f"Invalid nucleotides: {invalid}")
    return 81.5 + 16.6 * math.log10(na) + 0.41 * gc_percent(seq) - 600 / len(seq)


S = "ACGCGTGCCA"
NA = 0.05  # 50 mM

print(f"Sequence: {S}")
print(f"Length = {len(S)}, %GC = {gc_percent(S):.1f}%, [Na+] = {NA} M")
print()
print(f"Formula 1 (basic):         Tm = {melting_temperature(S)} °C")
print(f"Formula 2 (salt-adjusted): Tm = {melting_temperature_salt(S, NA):.2f} °C")
```
